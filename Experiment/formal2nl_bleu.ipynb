```python
import json
import numpy as np
import pandas as pd
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

# ---------- Load JSON ----------
json_path = "/Users/hmmoore/Desktop/AutoFormalization/Data/old/gpt_translations_latex_1_zero_shot_formatted.json"  # <- set your path
with open(json_path, "r") as f:
    data = json.load(f)

smooth_fn = SmoothingFunction().method1

def bleu_score(ref, cand):
    return sentence_bleu([ref.split()], cand.split(), smoothing_function=smooth_fn)

def capped_max(seq, k):
    k = min(k, len(seq))
    return max(seq[:k]) if k > 0 else 0.0

def capped_argmax(seq, k):
    k = min(k, len(seq))
    return int(np.argmax(seq[:k])) if k > 0 else -1

rows = []
for key, item in data.items():
    ref = (item.get("formal_s") or "").strip()

    cands = item.get("gpt_formal_s_list") or []
    cands = [c.strip() for c in cands if c.strip()]

    scores = [bleu_score(ref, c) for c in cands]

    # Pass@K from ORIGINAL order (no sorting!)
    p1  = scores[0] if len(scores) >= 1 else 0.0
    p2  = capped_max(scores, 2)
    p3  = capped_max(scores, 3)
    p5  = capped_max(scores, 5)
    p10 = capped_max(scores, 10)

    # Best overall (for reporting)
    best_idx = int(np.argmax(scores)) if scores else -1
    best_bleu = scores[best_idx] if best_idx >= 0 else 0.0
    best_cand = cands[best_idx] if best_idx >= 0 else ""

    # (Optional) which candidate achieves each Pass@K
    p1_idx  = 0 if len(scores) >= 1 else -1
    p2_idx  = capped_argmax(scores, 2)
    p3_idx  = capped_argmax(scores, 3)
    p5_idx  = capped_argmax(scores, 5)
    p10_idx = capped_argmax(scores, 10)

    rows.append({
        "id": key,
        "best_bleu": best_bleu,
        "best_candidate": best_cand,
        "pass@1": p1,
        "pass@2": p2,
        "pass@3": p3,
        "pass@5": p5,
        "pass@10": p10,
        "pass@1_idx": p1_idx,
        "pass@3_idx": p3_idx,
        "pass@5_idx": p5_idx,
        "pass@10_idx": p10_idx,
    })

df = pd.DataFrame(rows).set_index("id")
display(df.head())

print("\nAverage Pass@K across dataset:")
print(df[["pass@1","pass@2","pass@3","pass@5","pass@10"]].mean())
```
